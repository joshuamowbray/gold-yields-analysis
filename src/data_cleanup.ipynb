```python
import pandas as pd


yields = pd.read_csv('../data/raw/daily-yields.csv')
gold = pd.read_csv('../data/raw/monthly-gold.csv')

gold = gold.rename(columns = {'Price':'Gold Price'})
yields = yields.rename(columns = {'DGS10':'US10Y yields'})
yields = yields.rename(columns = {'observation_date':'Date'})


#Convert date values to datetime
yields['Date'] = pd.to_datetime(yields['Date'])
gold['Date'] = pd.to_datetime(gold['Date'])


#Calculate monthly mean for daily data and make dates allign
yields= yields.resample('ME', on='Date').mean()
gold = gold.resample('ME', on='Date').last()


#Calculate change
gold['Gold Price (% Change)'] = gold['Gold Price'].pct_change()*100
yields['US10Y yields (bp Change)'] = yields['US10Y yields'].diff()*100


#Gold and yields into one larger dataframe
comparison = gold.join(yields, how = 'inner')


comparison.to_csv('../data/processed/full_table.csv')
```
